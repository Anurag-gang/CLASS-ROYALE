This notebook demonstrates the **dataset-agnostic** architecture of the Class Royale ML Engine:

```python
from datetime import datetime, timezone
from app.schemas.interaction import StudentInteraction
from app.preprocessing.feature_pipeline import FeatureExtractor
from app.models.mastery_predictor import MasteryPredictor, MasteryBaseline
from app.models.question_recommender import QuestionRecommender
from app.schemas.recommendation import QuestionCandidate
from app.schemas.common import DifficultyLevel

# 1. Instantiate Canonical Student Interaction Telemetry
interaction = StudentInteraction(
    student_id="student_001",
    topic_id="binary-search-trees",
    question_id="q_bst_101",
    attempt_number=1,
    answer_correct=True,
    time_taken_seconds=42.0,
    hints_used=0,
    source_type="SYNTHETIC"
)
print("Canonical interaction:", interaction.model_dump())

# Verify target leakage stripping
print("Feature dict (targets excluded):", interaction.to_feature_dict(exclude_targets=True))
```

```python
# 2. Feature Extraction
extractor = FeatureExtractor()
features = extractor.extract_features_from_history(
    history=[interaction],
    target_topic_id="binary-search-trees",
    target_question_id="q_bst_102"
)
extractor.assert_no_leakage(features)
print("Engineered student features:")
for k, v in list(features.items())[:6]:
    print(f"  {k}: {v}")
```

```python
# 3. Pedagogical Recommendation Engine (Flow State Optimization)
recommender = QuestionRecommender()
candidates = [
    QuestionCandidate(question_id="q_bst_103", topic_id="binary-search-trees", difficulty=0.55, difficulty_level=DifficultyLevel.MEDIUM),
    QuestionCandidate(question_id="q_bst_104", topic_id="binary-search-trees", difficulty=0.85, difficulty_level=DifficultyLevel.HARD),
    QuestionCandidate(question_id="q_arrays_01", topic_id="arrays", difficulty=0.50, difficulty_level=DifficultyLevel.MEDIUM),
]

ranked = recommender.rank_candidates(
    candidates=candidates,
    student_mastery=0.70,
    weak_topics=["binary-search-trees"],
    top_k=2
)
for q in ranked:
    print(f"Rank #{q.rank}: {q.question_id} (Score={q.score}) -> {q.rationale}")
```
