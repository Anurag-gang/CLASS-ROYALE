# EdNet-KT1 Exploratory Data Analysis## Class Royale Machine Learning Engine — Phase 2This notebook conducts exploratory analysis on the **EdNet-KT1** benchmark dataset published by Riiid! AI Research.Dataset properties:- 784,309 unique student interaction files- 13,169 unique questions with TOEIC part and skill tags- Millisecond-level interaction telemetry (elapsed solving time, timestamps)

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load normalized cohort of 500 students (88,235 interactions)
parquet_path = Path("../datasets/processed/ednet_kt1/ednet_kt1_normalized.parquet")
df = pd.read_parquet(parquet_path)
print(f"Loaded EdNet normalized cohort: {df.shape[0]:,} rows, {df.shape[1]} columns")
df.head()
```

### 1. Student Activity & Interaction Length Distribution

```python
interactions_per_student = df.groupby("student_id").size()
print("Interactions per student summary:")
print(interactions_per_student.describe())
print(f"Top 5 most active students:\n{interactions_per_student.sort_values(ascending=False).head(5)}")
```

### 2. Question Frequency & Exposure Distribution

```python
interactions_per_question = df.groupby("question_id").size()
print("Interactions per question summary:")
print(interactions_per_question.describe())
print(f"Unique questions encountered in cohort: {df['question_id'].nunique():,}")
```

### 3. Correctness & Response Latency Analysis

```python
print(f"Overall Cohort Accuracy: {df['answer_correct'].mean():.2%}")
print("Response Time (seconds) Statistics:")
print(df['time_taken_seconds'].describe(percentiles=[0.25, 0.50, 0.75, 0.90, 0.99]))
```

### 4. Topic (TOEIC Part) Distribution

```python
part_counts = df["topic_id"].value_counts()
part_acc = df.groupby("topic_id")["answer_correct"].agg(["count", "mean"])
part_acc["accuracy_%"] = (part_acc["mean"] * 100).round(2)
print("Interactions and Accuracy by Part:")
print(part_acc)
```
