# MathE Exploratory Data Analysis## Assessing Mathematics Learning in Higher EducationThis notebook analyzes the **MathE** higher education mathematics dataset from the **UCI Machine Learning Repository** (Dataset 1031).Dataset properties:- 9,546 student responses across 14 university mathematics topics- 372 undergraduate students- 833 mathematics assessment questions categorized by difficulty level (Basic / Advanced)

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load raw and normalized MathE data
mathe_path = Path("../datasets/processed/mathe/mathe_normalized.parquet")
df = pd.read_parquet(mathe_path)
print(f"Loaded MathE normalized data: {df.shape[0]:,} rows, {df.shape[1]} columns")
df.head()
```

### 1. Mathematics Topics & Subtopics Breakdown

```python
topic_dist = df.groupby("topic_id")["answer_correct"].agg(["count", "mean"])
topic_dist["accuracy_%"] = (topic_dist["mean"] * 100).round(2)
print("Topic Distribution and Pass Rates:")
print(topic_dist.sort_values(by="count", ascending=False))
```

### 2. Difficulty Level Analysis (Basic vs Advanced)

```python
diff_acc = df.groupby("question_difficulty")["answer_correct"].agg(["count", "mean"])
diff_acc["accuracy_%"] = (diff_acc["mean"] * 100).round(2)
print("Accuracy across Question Difficulty Bands (0.35 = Basic, 0.75 = Advanced):")
print(diff_acc)
```

### 3. Student Interaction Volume & Class Balance

```python
print("Class balance (0 = Incorrect, 1 = Correct):")
print(df["answer_correct"].value_counts(normalize=True))
print(f"Unique Students: {df['student_id'].nunique()}, Unique Questions: {df['question_id'].nunique()}")
```
